```python
def create_category_mapping(category_data):
    return {
        int(item['id']): item['snippet']['title']
        for item in category_data['items']
    }

india_map = create_category_mapping(india_categories)
usa_map = create_category_mapping(usa_categories)
germany_map = create_category_mapping(germany_categories)
gb_map = create_category_mapping(gb_categories)
mexico_map = create_category_mapping(mexico_categories)

print("India categories:", len(india_map))
print("USA categories:", len(usa_map))
print("Germany categories:", len(germany_map))
print("Great Britain categories:", len(gb_map))
print("Mexico categories:", len(mexico_map))
```

```python
def get_category_name(row):
    category_id = int(row['category_id'])

    if row['country'] == 'India':
        return india_map.get(category_id)
    elif row['country'] == 'United States':
        return usa_map.get(category_id)
    elif row['country'] == 'Germany':
        return germany_map.get(category_id)
    elif row['country'] == 'Great Britain':
        return gb_map.get(category_id)
    elif row['country'] == 'Mexico':
        return mexico_map.get(category_id)

df['category_name'] = df.apply(get_category_name, axis=1)

print("Category mapping completed!")
```

```python
from transformers import pipeline

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-xlm-roberta-base-sentiment",
    device=0
)

print("Sentiment model loaded on GPU successfully.")
```

```python
import time
import numpy as np

all_texts = df['text_for_sentiment'].tolist()

sentiment_labels = []
sentiment_scores = []

batch_size = 32

start_time = time.time()

for start in range(0, len(all_texts), batch_size):
    batch = all_texts[start:start + batch_size]

    results = sentiment_pipeline(
        batch,
        batch_size=batch_size,
        truncation=True,
        max_length=256
    )

    sentiment_labels.extend([r['label'] for r in results])
    sentiment_scores.extend([r['score'] for r in results])

    # Progress every 5,000 rows
    if (start + len(batch)) % 5000 < batch_size:
        processed = start + len(batch)
        elapsed = time.time() - start_time
        rate = processed / elapsed
        remaining = (len(all_texts) - processed) / rate

        print(
            f"Processed {processed:,}/{len(all_texts):,} "
            f"({processed / len(all_texts) * 100:.1f}%) | "
            f"Elapsed: {elapsed / 60:.1f} min | "
            f"Estimated remaining: {remaining / 60:.1f} min"
        )

df['sentiment'] = sentiment_labels
df['sentiment_score'] = sentiment_scores

print("\nSentiment analysis completed!")
print(df['sentiment'].value_counts())
```

```python
import sqlite3

# Create an in-memory SQLite database
conn = sqlite3.connect(':memory:')

# Columns needed for SQL analysis
sql_columns = [
    'video_id',
    'country',
    'category_name',
    'trending_date',
    'publish_time',
    'views',
    'likes',
    'dislikes',
    'comment_count',
    'sentiment',
    'sentiment_score'
]

# Create SQL table
df[sql_columns].to_sql(
    'youtube_trending',
    conn,
    if_exists='replace',
    index=False
)

print("SQL table created successfully.")

# Check number of rows
result = pd.read_sql_query(
    "SELECT COUNT(*) AS total_rows FROM youtube_trending",
    conn
)

print(result)
```

```python
query = """
SELECT
    category_name,
    COUNT(*) AS video_records,
    ROUND(AVG(views), 0) AS average_views,
    SUM(views) AS total_views
FROM youtube_trending
GROUP BY category_name
ORDER BY average_views DESC;
"""

category_views = pd.read_sql_query(query, conn)

category_views
```

```python
query = """
SELECT
    country,
    category_name,
    COUNT(*) AS video_records,
    ROUND(AVG(views), 0) AS average_views
FROM youtube_trending
GROUP BY country, category_name
HAVING COUNT(*) >= 100
ORDER BY country, average_views DESC;
"""

country_category_views = pd.read_sql_query(query, conn)

country_category_views
```

```python
query = """
WITH ranked_categories AS (
    SELECT
        country,
        category_name,
        video_records,
        average_views,
        ROW_NUMBER() OVER (
            PARTITION BY country
            ORDER BY average_views DESC
        ) AS rank
    FROM (
        SELECT
            country,
            category_name,
            COUNT(*) AS video_records,
            ROUND(AVG(views), 0) AS average_views
        FROM youtube_trending
        GROUP BY country, category_name
        HAVING COUNT(*) >= 100
    )
)

SELECT
    country,
    category_name,
    video_records,
    average_views
FROM ranked_categories
WHERE rank = 1
ORDER BY country;
"""

top_category_by_country = pd.read_sql_query(query, conn)

top_category_by_country
```

```python
query = """
SELECT
    video_id,
    country,
    category_name,
    MIN(trending_date) AS first_trending_date,
    MAX(trending_date) AS last_trending_date,
    COUNT(*) AS trending_days
FROM youtube_trending
WHERE video_id IS NOT NULL
GROUP BY video_id, country, category_name
HAVING COUNT(*) >= 2
ORDER BY trending_days DESC;
"""

trending_duration = pd.read_sql_query(query, conn)

trending_duration.head(20)
```

```python
query = """
WITH video_duration AS (
    SELECT
        video_id,
        country,
        COUNT(*) AS trending_days
    FROM youtube_trending
    WHERE video_id IS NOT NULL
    GROUP BY video_id, country
    HAVING COUNT(*) >= 2
)

SELECT
    country,
    COUNT(*) AS videos_tracked,
    ROUND(AVG(trending_days), 2) AS average_trending_days,
    MAX(trending_days) AS maximum_trending_days
FROM video_duration
GROUP BY country
ORDER BY average_trending_days DESC;
"""

duration_by_country = pd.read_sql_query(query, conn)

duration_by_country
```

```python
query = """
SELECT
    trending_date,
    country,
    COUNT(*) AS trending_records
FROM youtube_trending
GROUP BY trending_date, country
ORDER BY trending_date, country;
"""

trending_over_time = pd.read_sql_query(query, conn)

trending_over_time.head(20)
```

```python
query = """
SELECT
    strftime('%Y-%m', trending_date) AS month,
    country,
    COUNT(DISTINCT video_id) AS unique_trending_videos
FROM youtube_trending
WHERE video_id IS NOT NULL
GROUP BY month, country
ORDER BY month, country;
"""

monthly_trending = pd.read_sql_query(query, conn)

monthly_trending.head(25)
```

```python
query = """
SELECT
    country,
    sentiment,
    COUNT(*) AS video_records,
    ROUND(AVG(views), 0) AS average_views,
    ROUND(AVG(sentiment_score), 3) AS average_confidence
FROM youtube_trending
GROUP BY country, sentiment
ORDER BY country, average_views DESC;
"""

sentiment_by_country = pd.read_sql_query(query, conn)

sentiment_by_country
```

```python
query = """
SELECT
    country,
    sentiment,
    COUNT(*) AS video_records,
    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (PARTITION BY country),
        2
    ) AS percentage_of_records
FROM youtube_trending
GROUP BY country, sentiment
ORDER BY country, percentage_of_records DESC;
"""

sentiment_distribution = pd.read_sql_query(query, conn)

sentiment_distribution
```
